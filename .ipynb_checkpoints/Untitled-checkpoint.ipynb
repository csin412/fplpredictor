```python
def top_n_hit_rate(df, n=10):
    hits = []
    for (season, gw), group in df.groupby(['season', 'round']):
        actual_top = set(group.nlargest(n, 'total_points')['name'])
        predicted_top = set(group.nlargest(n, 'rolling_5_avg')['name'])
        overlap = len(actual_top & predicted_top)
        hits.append(overlap / n)
    return sum(hits)/len(hits)

hit_rate = top_n_hit_rate(relevant_eval)
print(f"Average overlap in top 10: {hit_rate:.1%}")
```

```python
def hit_defcon_threshold(row):
    if pd.isna(row['defensive_contribution']):
        return None
    if row['position'] == 'DEF':
        return 1 if row['defensive_contribution'] >= 10 else 0
    if row['position'] == 'MID':
        return 1 if row['defensive_contribution'] >= 12 else 0
    else:
        return 0

full_df['hit_defcon'] = full_df.apply(hit_defcon_threshold, axis=1)

full_df['rolling_5_defcon_rate'] = full_df.groupby('name')['hit_defcon'].transform(lambda x: x.rolling(window=5, min_periods=1).mean().shift(1))
```

```python
# --- Load all seasons of odds data ---
odds_seasons = ['2122', '2223', '2324', '2425', '2526']
season_label_map = dict(zip(odds_seasons, seasons))  # maps '2122' -> '2021-22' etc

odds_dfs = []
for s in odds_seasons:
    path = f"C:\\Users\\charl\\Football-Data\\odds{s}.csv"
    d = pd.read_csv(path)
    d['season'] = season_label_map[s]
    odds_dfs.append(d)

all_odds = pd.concat(odds_dfs, ignore_index=True).copy()
all_odds['Date'] = pd.to_datetime(all_odds['Date'], dayfirst=True)
print(all_odds.shape)
```

```python
# Fill this in once you've compared the two lists above
team_name_map = {
    'Man United': 'Man Utd',
    'Tottenham': 'Spurs',
    'Man Utd': 'Man United',
    'Spurs': 'Tottenham',
    'Sheffield United': 'Sheffield Utd'
}

all_odds['HomeTeam_mapped'] = all_odds['HomeTeam'].replace(team_name_map)
all_odds['AwayTeam_mapped'] = all_odds['AwayTeam'].replace(team_name_map)
```

```python
print(sorted(all_odds['HomeTeam_mapped'].unique()))
print(sorted(full_df['opponent_team_name'].unique()))
```

```python
# --- Convert odds to devigged implied probabilities ---
implied_H = 1 / all_odds['AvgH']
implied_D = 1 / all_odds['AvgD']
implied_A = 1 / all_odds['AvgA']
overround = implied_H + implied_D + implied_A

all_odds['prob_home_win'] = implied_H / overround
all_odds['prob_draw']     = implied_D / overround
all_odds['prob_away_win'] = implied_A / overround

implied_over = 1 / all_odds['Avg>2.5']
implied_under = 1 / all_odds['Avg<2.5']
ou_overround = implied_over + implied_under
all_odds['prob_over_2.5'] = implied_over / ou_overround
```

```python
# --- Reshape to one row per team per match (home perspective + away perspective) ---
home_rows = all_odds[['season', 'Date', 'HomeTeam_mapped', 'AwayTeam_mapped',
                       'prob_home_win', 'prob_draw', 'prob_away_win', 'prob_over_2.5']].copy()
home_rows = home_rows.rename(columns={
    'HomeTeam_mapped': 'team', 'AwayTeam_mapped': 'opponent_team_name',
    'prob_home_win': 'team_win_prob', 'prob_away_win': 'opponent_win_prob'
})
home_rows['was_home'] = True

away_rows = all_odds[['season', 'Date', 'HomeTeam_mapped', 'AwayTeam_mapped',
                       'prob_home_win', 'prob_draw', 'prob_away_win', 'prob_over_2.5']].copy()
away_rows = away_rows.rename(columns={
    'AwayTeam_mapped': 'team', 'HomeTeam_mapped': 'opponent_team_name',
    'prob_away_win': 'team_win_prob', 'prob_home_win': 'opponent_win_prob'
})
away_rows['was_home'] = False

team_odds = pd.concat([home_rows, away_rows], ignore_index=True)
team_odds = team_odds.rename(columns={'prob_draw': 'team_draw_prob', 'prob_over_2.5': 'match_over_2.5_prob'})
```

```python
# --- Merge into full_df on season + team + opponent + was_home ---
# kickoff_time in full_df needs to be date-matched to Date in team_odds
full_df['kickoff_date'] = pd.to_datetime(full_df['kickoff_time']).dt.date
team_odds['Date_only'] = team_odds['Date'].dt.date
```

```python
from scipy.stats import poisson
from scipy.optimize import brentq

def implied_total_goals(prob_over_2_5):
    # find lambda such that P(Poisson(lambda) > 2.5) = prob_over_2_5
    if pd.isna(prob_over_2_5):
        return np.nan
    def f(lam):
        return poisson.sf(2, lam) - prob_over_2_5  # sf(2, lam) = P(X >= 3) = P(X > 2.5)
    try:
        return brentq(f, 0.1, 8.0)
    except ValueError:
        return np.nan

all_odds['match_expected_goals'] = all_odds['prob_over_2.5'].apply(implied_total_goals)
```

```python
all_odds['home_goal_share'] = all_odds['prob_home_win'] + 0.5 * all_odds['prob_draw']
all_odds['away_goal_share'] = all_odds['prob_away_win'] + 0.5 * all_odds['prob_draw']

all_odds['home_expected_goals'] = all_odds['match_expected_goals'] * all_odds['home_goal_share']
all_odds['away_expected_goals'] = all_odds['match_expected_goals'] * all_odds['away_goal_share']
```

```python
home_rows = all_odds[['season', 'Date', 'HomeTeam_mapped', 'AwayTeam_mapped',
                       'prob_home_win', 'prob_draw', 'prob_away_win',
                       'home_expected_goals', 'away_expected_goals']].copy()
home_rows = home_rows.rename(columns={
    'HomeTeam_mapped': 'team', 'AwayTeam_mapped': 'opponent_team_name',
    'prob_home_win': 'team_win_prob', 'prob_away_win': 'opponent_win_prob',
    'home_expected_goals': 'market_expected_goals_for',
    'away_expected_goals': 'market_expected_goals_against'
})
home_rows['was_home'] = True

away_rows = all_odds[['season', 'Date', 'HomeTeam_mapped', 'AwayTeam_mapped',
                       'prob_home_win', 'prob_draw', 'prob_away_win',
                       'home_expected_goals', 'away_expected_goals']].copy()
away_rows = away_rows.rename(columns={
    'AwayTeam_mapped': 'team', 'HomeTeam_mapped': 'opponent_team_name',
    'prob_away_win': 'team_win_prob', 'prob_home_win': 'opponent_win_prob',
    'away_expected_goals': 'market_expected_goals_for',
    'home_expected_goals': 'market_expected_goals_against'
})
away_rows['was_home'] = False

team_odds = pd.concat([home_rows, away_rows], ignore_index=True).copy()
team_odds = team_odds.rename(columns={'prob_draw': 'team_draw_prob'})
team_odds['Date_only'] = team_odds['Date'].dt.date

full_df = full_df.merge(
    team_odds[['season', 'Date_only', 'team', 'opponent_team_name', 'was_home',
               'team_win_prob', 'team_draw_prob', 'opponent_win_prob',
               'market_expected_goals_for', 'market_expected_goals_against']],
    left_on=['season', 'kickoff_date', 'team', 'opponent_team_name', 'was_home'],
    right_on=['season', 'Date_only', 'team', 'opponent_team_name', 'was_home'],
    how='left'
)
print(full_df['market_expected_goals_for'].isna().sum(), "rows failed to match odds")
```

```python
def precision_at_n(df, pred_col, n=10, threshold=6):
    precisions = []
    for (season, gw), group in df.groupby(['season', 'round']):
        top_n = group.nlargest(n, pred_col)
        hits = (top_n['total_points'] >= threshold).sum()
        precisions.append(hits / n)
    return sum(precisions) / len(precisions)

for n in [5, 10, 15]:
    p = precision_at_n(test_relevant_odds_copy, 'model_pred', n=n)
    print(f"Precision@{n}: {p:.1%}")
```

```python
import requests

def fetch_current_odds(api_key):
    url = "https://api.the-odds-api.com/v4/sports/soccer_epl/odds"
    params = {
        "apiKey": api_key,
        "regions": "uk",
        "markets": "h2h,totals",   # h2h = match winner, totals = over/under goals
        "oddsFormat": "decimal",
        "dateFormat": "iso"
    }
    response = requests.get(url, params=params)
    print(f"Requests remaining this month: {response.headers.get('x-requests-remaining')}")
    return response.json()

live_odds_raw = fetch_current_odds(ODDS_API_KEY)
print(len(live_odds_raw), "fixtures returned")
```

```python
def parse_live_odds(raw_odds):
    rows = []
    for match in raw_odds:
        home_team = match['home_team']
        away_team = match['away_team']
        commence_time = match['commence_time']

        h_odds, d_odds, a_odds = [], [], []
        over_odds, under_odds = [], []

        for bookmaker in match['bookmakers']:
            for market in bookmaker['markets']:
                if market['key'] == 'h2h':
                    for outcome in market['outcomes']:
                        if outcome['name'] == home_team:
                            h_odds.append(outcome['price'])
                        elif outcome['name'] == away_team:
                            a_odds.append(outcome['price'])
                        else:
                            d_odds.append(outcome['price'])  # "Draw"
                elif market['key'] == 'totals':
                    for outcome in market['outcomes']:
                        if outcome['point'] == 2.5:
                            if outcome['name'] == 'Over':
                                over_odds.append(outcome['price'])
                            elif outcome['name'] == 'Under':
                                under_odds.append(outcome['price'])

        rows.append({
            'HomeTeam': home_team,
            'AwayTeam': away_team,
            'Date': commence_time,
            'AvgH': sum(h_odds)/len(h_odds) if h_odds else None,
            'AvgD': sum(d_odds)/len(d_odds) if d_odds else None,
            'AvgA': sum(a_odds)/len(a_odds) if a_odds else None,
            'Avg>2.5': sum(over_odds)/len(over_odds) if over_odds else None,
            'Avg<2.5': sum(under_odds)/len(under_odds) if under_odds else None,
        })
    return pd.DataFrame(rows)

live_odds_df = parse_live_odds(live_odds_raw)
print(live_odds_df)
```

```python
implied_H = 1 / live_odds_df['AvgH']
implied_D = 1 / live_odds_df['AvgD']
implied_A = 1 / live_odds_df['AvgA']
overround = implied_H + implied_D + implied_A

live_odds_df['prob_home_win'] = implied_H / overround
live_odds_df['prob_draw'] = implied_D / overround
live_odds_df['prob_away_win'] = implied_A / overround

implied_over = 1 / live_odds_df['Avg>2.5']
implied_under = 1 / live_odds_df['Avg<2.5']
ou_overround = implied_over + implied_under
live_odds_df['prob_over_2.5'] = implied_over / ou_overround
live_odds_df['match_expected_goals'] = live_odds_df['prob_over_2.5'].apply(implied_total_goals)

live_odds_df['home_goal_share'] = live_odds_df['prob_home_win'] + 0.5 * live_odds_df['prob_draw']
live_odds_df['away_goal_share'] = live_odds_df['prob_away_win'] + 0.5 * live_odds_df['prob_draw']
live_odds_df['home_expected_goals'] = live_odds_df['match_expected_goals'] * live_odds_df['home_goal_share']
live_odds_df['away_expected_goals'] = live_odds_df['match_expected_goals'] * live_odds_df['away_goal_share']
```

```python
home_rows_live = live_odds_df[['HomeTeam_mapped', 'AwayTeam_mapped',
                                'prob_home_win', 'prob_draw', 'prob_away_win',
                                'home_expected_goals', 'away_expected_goals']].copy()
home_rows_live = home_rows_live.rename(columns={
    'HomeTeam_mapped': 'team', 'AwayTeam_mapped': 'opponent_team_name',
    'prob_home_win': 'team_win_prob', 'prob_away_win': 'opponent_win_prob',
    'home_expected_goals': 'market_expected_goals_for',
    'away_expected_goals': 'market_expected_goals_against'
})
home_rows_live['was_home'] = True

away_rows_live = live_odds_df[['HomeTeam_mapped', 'AwayTeam_mapped',
                                'prob_home_win', 'prob_draw', 'prob_away_win',
                                'home_expected_goals', 'away_expected_goals']].copy()
away_rows_live = away_rows_live.rename(columns={
    'AwayTeam_mapped': 'team', 'HomeTeam_mapped': 'opponent_team_name',
    'prob_away_win': 'team_win_prob', 'prob_home_win': 'opponent_win_prob',
    'away_expected_goals': 'market_expected_goals_for',
    'home_expected_goals': 'market_expected_goals_against'
})
away_rows_live['was_home'] = False

live_team_odds = pd.concat([home_rows_live, away_rows_live], ignore_index=True)
live_team_odds = live_team_odds.rename(columns={'prob_draw': 'team_draw_prob'})
live_team_odds['market_cs_prob'] = np.exp(-live_team_odds['market_expected_goals_against'])
```

```python
stale_odds_cols = ['team_win_prob', 'team_draw_prob', 'opponent_win_prob',
                    'market_expected_goals_for', 'market_expected_goals_against', 'market_cs_prob']

next_week_players = next_week_players.drop(columns=[c for c in stale_odds_cols if c in next_week_players.columns])

next_week_players = next_week_players.merge(
    live_team_odds[['team', 'opponent_team_name', 'was_home',
                     'team_win_prob', 'team_draw_prob', 'opponent_win_prob',
                     'market_expected_goals_for', 'market_expected_goals_against', 'market_cs_prob']],
    on=['team', 'opponent_team_name', 'was_home'],
    how='left'
)

print(next_week_players['team_win_prob'].isna().sum(), "players missing odds data")
print(next_week_players[next_week_players['team_win_prob'].isna()]['team'].unique())
```
